```python
# Tracker 02.04 — Salesperson Master Generator

import pandas as pd
import numpy as np
from faker import Faker
import random
from datetime import datetime, timedelta
```

```python
# Initialize Faker and random seed

fake = Faker("en_IN")

Faker.seed(42)
random.seed(42)
np.random.seed(42)

NUM_SALESPERSONS = 500

print("Faker initialized successfully.")
print("Salespersons to generate:", NUM_SALESPERSONS)
```

```python
# Generate Salesperson IDs

salesperson_ids = [
    f"SP{str(i).zfill(5)}"
    for i in range(1, NUM_SALESPERSONS + 1)
]

print("Total Salesperson IDs:", len(salesperson_ids))
print("First 5 IDs:", salesperson_ids[:5])
print("Last 5 IDs:", salesperson_ids[-5:])
```

```python
# Load Store Master and define salesperson rules

stores_df = pd.read_csv("../data/synthetic/stores_master.csv")

roles = [
    "Sales Associate",
    "Senior Sales Associate",
    "Sales Executive",
    "Senior Sales Executive"
]

status_options = [
    "Active",
    "Inactive"
]

print("Stores loaded:", len(stores_df))
print("Available roles:", roles)
print("Status options:", status_options)
```

```python
# Generate salesperson records

salespersons = []

for salesperson_id in salesperson_ids:

    # Select a store
    store_id = random.choice(stores_df["store_id"].tolist())

    # Get store details for region linkage
    store_row = stores_df[stores_df["store_id"] == store_id].iloc[0]

    joining_date = fake.date_between(
        start_date="-8y",
        end_date="today"
    )

    role = random.choices(
        roles,
        weights=[50, 25, 20, 5],
        k=1
    )[0]

    status = random.choices(
        status_options,
        weights=[90, 10],
        k=1
    )[0]

    salespersons.append({
        "salesperson_id": salesperson_id,
        "salesperson_name": fake.name(),
        "store_id": store_id,
        "joining_date": joining_date,
        "role": role,
        "status": status
    })

print("Total salesperson records generated:", len(salespersons))
```

```python
# Convert salesperson records into a Pandas DataFrame

salespersons_df = pd.DataFrame(salespersons)

print("DataFrame Shape:", salespersons_df.shape)
print("Total Salespersons:", len(salespersons_df))
print("Total Columns:", len(salespersons_df.columns))

salespersons_df.head()
```

```python
# Validate salesperson data

print("Missing values:")
print(salespersons_df.isnull().sum())

print("\nDuplicate Salesperson IDs:")
print(salespersons_df["salesperson_id"].duplicated().sum())

print("\nSalesperson ID uniqueness:")
print(salespersons_df["salesperson_id"].nunique())

print("\nStore ID linkage:")
print(salespersons_df["store_id"].isin(stores_df["store_id"]).value_counts())

print("\nSalesperson Status:")
print(salespersons_df["status"].value_counts())

print("\nSalesperson Role:")
print(salespersons_df["role"].value_counts())
```

```python
# Save Salesperson Master dataset

from pathlib import Path

output_path = Path("../data/synthetic/salespersons_master.csv")

output_path.parent.mkdir(parents=True, exist_ok=True)

salespersons_df.to_csv(output_path, index=False)

print("Salesperson Master saved successfully.")
print("File path:", output_path)
print("Rows saved:", len(salespersons_df))
```

```python
# Confirm saved Salesperson Master file

saved_df = pd.read_csv(output_path)

print("Saved file loaded successfully.")
print("Shape:", saved_df.shape)
print("Columns:", list(saved_df.columns))
```

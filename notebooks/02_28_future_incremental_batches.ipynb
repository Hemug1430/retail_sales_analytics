```python
from pathlib import Path
import pandas as pd

# ============================================================
# 02.28 - FUTURE INCREMENTAL BATCHES
# STEP 2 - PREPARE FUTURE BATCH STRUCTURE
# ============================================================

PROJECT_ROOT = Path.cwd().parent
SALES_DIR = PROJECT_ROOT / "data" / "synthetic"

OUTPUT_DIR = SALES_DIR / "future_incremental_batches"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("=" * 70)
print("02.28 STEP 2 - PREPARE FUTURE INCREMENTAL BATCH STRUCTURE")
print("=" * 70)

# ------------------------------------------------------------
# 1. Define future incremental batch months
# ------------------------------------------------------------

future_periods = pd.period_range(
    start="2026-01",
    end="2026-08",
    freq="M"
)

future_batches = []

for batch_number, period in enumerate(future_periods, start=1):

    future_batches.append({
        "batch_number": batch_number,
        "batch_month": str(period),
        "batch_file_name": f"incremental_sales_{period}.csv"
    })

future_batches_df = pd.DataFrame(future_batches)

# ------------------------------------------------------------
# 2. Display planned batches
# ------------------------------------------------------------

print("\nFUTURE INCREMENTAL BATCHES")
print("-" * 70)

print(future_batches_df.to_string(index=False))

# ------------------------------------------------------------
# 3. Validate batch count
# ------------------------------------------------------------

expected_batch_count = 8

batch_count_valid = (
    len(future_batches_df) == expected_batch_count
)

# ------------------------------------------------------------
# 4. Validate monthly sequence
# ------------------------------------------------------------

expected_future_months = [
    str(period)
    for period in future_periods
]

actual_future_months = (
    future_batches_df["batch_month"].tolist()
)

month_sequence_valid = (
    actual_future_months == expected_future_months
)

# ------------------------------------------------------------
# 5. Validate unique batch numbers
# ------------------------------------------------------------

unique_batch_numbers = (
    future_batches_df["batch_number"].is_unique
)

# ------------------------------------------------------------
# 6. Validate output folder
# ------------------------------------------------------------

output_folder_valid = OUTPUT_DIR.exists()

# ------------------------------------------------------------
# 7. Save batch plan
# ------------------------------------------------------------

BATCH_PLAN_FILE = (
    OUTPUT_DIR / "future_incremental_batch_plan.csv"
)

future_batches_df.to_csv(
    BATCH_PLAN_FILE,
    index=False
)

# ------------------------------------------------------------
# 8. Final validation
# ------------------------------------------------------------

print("\nVALIDATION")
print("-" * 70)

print("Expected batches :", expected_batch_count)
print("Actual batches   :", len(future_batches_df))
print("Batch count valid:", batch_count_valid)

print("\nExpected first month:", expected_future_months[0])
print("Expected last month :", expected_future_months[-1])
print("Monthly sequence valid:", month_sequence_valid)

print("\nUnique batch numbers:", unique_batch_numbers)
print("Output folder exists:", output_folder_valid)

print("\nBatch plan saved to:")
print(BATCH_PLAN_FILE)

if (
    batch_count_valid
    and month_sequence_valid
    and unique_batch_numbers
    and output_folder_valid
    and BATCH_PLAN_FILE.exists()
):
    print("\n02.28 Step 2 preparation: PASSED")
else:
    print("\n02.28 Step 2 preparation: NEEDS REVIEW")
```

```python
from pathlib import Path
import pandas as pd
import shutil

# ============================================================
# 02.28 - FUTURE INCREMENTAL BATCHES
# STEP 3 - CREATE BATCH 1: JANUARY 2026
# ============================================================

PROJECT_ROOT = Path.cwd().parent
SALES_DIR = PROJECT_ROOT / "data" / "synthetic"

OUTPUT_DIR = (
    SALES_DIR
    / "future_incremental_batches"
    / "batch_01_2026-01"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("=" * 70)
print("02.28 STEP 3 - CREATE BATCH 1: JANUARY 2026")
print("=" * 70)

# ------------------------------------------------------------
# 1. Source files
# ------------------------------------------------------------

SALES_SOURCE = SALES_DIR / "sales_2026-01.csv"
RETURNS_SOURCE = SALES_DIR / "returns_2026-01.csv"
TARGETS_SOURCE = SALES_DIR / "targets.csv"

# ------------------------------------------------------------
# 2. Validate sales and returns source files
# ------------------------------------------------------------

print("\nSOURCE FILE VALIDATION")
print("-" * 70)

print("Sales file exists   :", SALES_SOURCE.exists())
print("Returns file exists :", RETURNS_SOURCE.exists())
print("Targets file exists :", TARGETS_SOURCE.exists())

if not SALES_SOURCE.exists():
    raise FileNotFoundError(f"Missing source file: {SALES_SOURCE}")

if not RETURNS_SOURCE.exists():
    raise FileNotFoundError(f"Missing source file: {RETURNS_SOURCE}")

if not TARGETS_SOURCE.exists():
    raise FileNotFoundError(f"Missing source file: {TARGETS_SOURCE}")

# ------------------------------------------------------------
# 3. Copy January sales file
# ------------------------------------------------------------

sales_output = OUTPUT_DIR / "sales_2026-01.csv"

shutil.copy2(
    SALES_SOURCE,
    sales_output
)

# ------------------------------------------------------------
# 4. Copy January returns file
# ------------------------------------------------------------

returns_output = OUTPUT_DIR / "returns_2026-01.csv"

shutil.copy2(
    RETURNS_SOURCE,
    returns_output
)

# ------------------------------------------------------------
# 5. Extract January 2026 targets
# ------------------------------------------------------------

targets_df = pd.read_csv(TARGETS_SOURCE)

targets_df["target_month"] = pd.to_datetime(
    targets_df["target_month"]
)

jan_targets = targets_df[
    targets_df["target_month"].dt.to_period("M")
    == pd.Period("2026-01")
].copy()

targets_output = OUTPUT_DIR / "targets_2026-01.csv"

jan_targets.to_csv(
    targets_output,
    index=False
)

# ------------------------------------------------------------
# 6. Inspect created batch
# ------------------------------------------------------------

print("\nBATCH 1 OUTPUT")
print("-" * 70)

print("Output folder:")
print(OUTPUT_DIR)

print("\nCreated files:")

for file in sorted(OUTPUT_DIR.glob("*.csv")):
    print(
        f"- {file.name} | "
        f"{file.stat().st_size:,} bytes"
    )

# ------------------------------------------------------------
# 7. Row counts
# ------------------------------------------------------------

sales_rows = pd.read_csv(
    sales_output,
    usecols=["sales_line_id"]
).shape[0]

returns_rows = pd.read_csv(
    returns_output
).shape[0]

target_rows = len(jan_targets)

print("\nROW COUNTS")
print("-" * 70)

print("January sales rows   :", sales_rows)
print("January returns rows :", returns_rows)
print("January target rows  :", target_rows)

# ------------------------------------------------------------
# 8. Final validation
# ------------------------------------------------------------

required_outputs = [
    sales_output,
    returns_output,
    targets_output
]

all_outputs_exist = all(
    file.exists()
    for file in required_outputs
)

target_month_valid = (
    not jan_targets.empty
    and (
        jan_targets["target_month"]
        .dt.to_period("M")
        .eq(pd.Period("2026-01"))
        .all()
    )
)

print("\nFINAL VALIDATION")
print("-" * 70)

print("All Batch 1 files created:", all_outputs_exist)
print("January target month valid:", target_month_valid)

if all_outputs_exist and target_month_valid:
    print("\n02.28 Step 3 batch creation: PASSED")
else:
    print("\n02.28 Step 3 batch creation: NEEDS REVIEW")
```

```python
# ============================================================
# 02.28 STEP 16 - FINAL VALIDATION
# ============================================================

from pathlib import Path
import pandas as pd

print("=" * 70)
print("02.28 STEP 16 - FINAL VALIDATION")
print("=" * 70)

BASE_DIR = Path(
    r"C:\Users\HP\OneDrive\Desktop\retail_sales_analytics"
)

BATCH_ROOT = (
    BASE_DIR
    / "data"
    / "synthetic"
    / "future_incremental_batches"
)

expected_months = [
    "2026-01",
    "2026-02",
    "2026-03",
    "2026-04",
    "2026-05",
    "2026-06",
    "2026-07",
    "2026-08"
]

results = []

for batch_number, month in enumerate(expected_months, start=1):

    batch_dir = (
        BATCH_ROOT
        / f"batch_{batch_number:02d}_{month}"
    )

    sales_file = batch_dir / f"sales_{month}.csv"
    returns_file = batch_dir / f"returns_{month}.csv"
    targets_file = batch_dir / f"targets_{month}.csv"

    sales_rows = 0
    returns_rows = 0
    target_rows = 0

    sales_valid = False
    returns_valid = False
    targets_valid = False

    # --------------------------------------------------------
    # SALES VALIDATION
    # --------------------------------------------------------

    if sales_file.exists():

        sales_df = pd.read_csv(sales_file)

        sales_rows = len(sales_df)

        sales_valid = (
            "order_date" in sales_df.columns
            and
            pd.to_datetime(
                sales_df["order_date"],
                errors="coerce"
            ).dt.to_period("M").eq(
                pd.Period(month)
            ).all()
        )

    # --------------------------------------------------------
    # RETURNS VALIDATION
    # --------------------------------------------------------

    if returns_file.exists():

        returns_df = pd.read_csv(returns_file)

        returns_rows = len(returns_df)

        returns_valid = (
            "return_date" in returns_df.columns
            and
            pd.to_datetime(
                returns_df["return_date"],
                errors="coerce"
            ).dt.to_period("M").eq(
                pd.Period(month)
            ).all()
        )

    # --------------------------------------------------------
    # TARGET VALIDATION
    # --------------------------------------------------------

    if targets_file.exists():

        targets_df = pd.read_csv(targets_file)

        target_rows = len(targets_df)

        targets_valid = (
            "target_month" in targets_df.columns
            and
            pd.to_datetime(
                targets_df["target_month"],
                errors="coerce"
            ).dt.to_period("M").eq(
                pd.Period(month)
            ).all()
        )

    # --------------------------------------------------------
    # BATCH VALIDATION
    # --------------------------------------------------------

    batch_valid = (
        batch_dir.exists()
        and sales_file.exists()
        and returns_file.exists()
        and targets_file.exists()
        and sales_rows > 0
        and returns_rows > 0
        and target_rows > 0
        and sales_valid
        and returns_valid
        and targets_valid
    )

    results.append({
        "batch": batch_number,
        "month": month,
        "sales_rows": sales_rows,
        "returns_rows": returns_rows,
        "target_rows": target_rows,
        "batch_valid": batch_valid
    })

# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

validation_df = pd.DataFrame(results)

print("\nBATCH VALIDATION SUMMARY")
print("-" * 70)

print(
    validation_df.to_string(index=False)
)

# ------------------------------------------------------------
# FINAL VALIDATION
# ------------------------------------------------------------

expected_batches = len(validation_df)

valid_batches = validation_df["batch_valid"].sum()

total_sales_rows = validation_df["sales_rows"].sum()

total_returns_rows = validation_df["returns_rows"].sum()

total_target_rows = validation_df["target_rows"].sum()

all_batches_valid = (
    valid_batches == expected_batches
)

print("\nFINAL VALIDATION")
print("-" * 70)

print(f"Expected batches       : {expected_batches}")
print(f"Valid batches          : {valid_batches}")
print(f"Total sales rows       : {total_sales_rows:,}")
print(f"Total returns rows     : {total_returns_rows:,}")
print(f"Total target rows      : {total_target_rows:,}")
print(f"All batches valid      : {all_batches_valid}")

if all_batches_valid:

    print("\n02.28 Step 16 validation: PASSED")
    print("Future incremental batch package validated successfully.")

else:

    print("\n02.28 Step 16 validation: NEEDS REVIEW")
```

```python
# Tracker 02.05 — Promotion Master Generator

import pandas as pd
import numpy as np
from faker import Faker
import random
from datetime import datetime, timedelta
from pathlib import Path


# Initialize Faker and random seed

fake = Faker("en_IN")

Faker.seed(42)
random.seed(42)
np.random.seed(42)

NUM_PROMOTIONS = 72

print("Faker initialized successfully.")
print("Promotions to generate:", NUM_PROMOTIONS)


# Generate Promotion IDs

promotion_ids = [
    f"PROMO{str(i).zfill(5)}"
    for i in range(1, NUM_PROMOTIONS + 1)
]

print("Total Promotion IDs:", len(promotion_ids))
print("First 5 IDs:", promotion_ids[:5])
print("Last 5 IDs:", promotion_ids[-5:])


# Define promotion rules

promotion_categories = [
    "Electronics",
    "Grocery",
    "Fashion",
    "Home & Kitchen",
    "Beauty & Personal Care"
]

promotion_channels = [
    "Online",
    "Store",
    "Both"
]

promotion_statuses = [
    "Planned",
    "Active",
    "Expired",
    "Cancelled"
]

print("Promotion categories:", promotion_categories)
print("Promotion channels:", promotion_channels)
print("Promotion statuses:", promotion_statuses)


# Load Product Master for promotion linkage

project_path = Path.cwd().parent

product_file = project_path / "data" / "synthetic" / "products_master.csv"

print("Product Master path:", product_file)
print("File exists:", product_file.exists())

if not product_file.exists():
    raise FileNotFoundError("products_master.csv was not found.")

products_df = pd.read_csv(product_file)

print("Products loaded:", len(products_df))
print("Product categories:", products_df["category"].nunique())
print("Available categories:")
print(products_df["category"].unique())


# Generate promotion records

promotions = []

campaign_start = datetime.strptime(
    "2025-01-01",
    "%Y-%m-%d"
).date()

campaign_end = datetime.strptime(
    "2026-08-31",
    "%Y-%m-%d"
).date()

for promotion_id in promotion_ids:

    promotion_name = f"{fake.company()} Retail Campaign"

    promotion_type = random.choice([
        "Seasonal",
        "Festive",
        "Clearance",
        "Flash Sale",
        "Loyalty"
    ])

    start_date = fake.date_between(
        start_date=campaign_start,
        end_date=campaign_end
    )

    campaign_days = random.randint(7, 45)

    end_date = start_date + timedelta(days=campaign_days)

    if end_date > campaign_end:
        end_date = campaign_end

    applicable_category = random.choice(
        promotion_categories
    )

    applicable_channel = random.choice(
        promotion_channels
    )

    discount_pct = round(
        np.random.uniform(5, 30),
        2
    )

    minimum_order_value = random.choice([
        0,
        500,
        1000,
        1500,
        2000,
        3000,
        5000
    ])

    if start_date > datetime.today().date():
        status = "Planned"
    elif end_date >= datetime.today().date():
        status = "Active"
    else:
        status = "Expired"

    promotions.append({
        "promotion_id": promotion_id,
        "promotion_name": promotion_name,
        "promotion_type": promotion_type,
        "start_date": start_date,
        "end_date": end_date,
        "discount_pct": discount_pct,
        "applicable_category": applicable_category,
        "applicable_channel": applicable_channel,
        "minimum_order_value": minimum_order_value,
        "status": status
    })

print(
    "Total promotion records generated:",
    len(promotions)
)


# Convert promotion records into a Pandas DataFrame

promotions_df = pd.DataFrame(promotions)

print("DataFrame Shape:", promotions_df.shape)
print("Total Promotions:", len(promotions_df))
print("Total Columns:", len(promotions_df.columns))

promotions_df.head()


# Save Promotion Master dataset

output_path = Path(
    "../data/synthetic/promotions_master.csv"
)

output_path.parent.mkdir(
    parents=True,
    exist_ok=True
)

promotions_df.to_csv(
    output_path,
    index=False
)

print("Promotion Master saved successfully.")
print("File path:", output_path)
print("Rows saved:", len(promotions_df))


# Confirm saved Promotion Master file

saved_df = pd.read_csv(output_path)

print("Saved file loaded successfully.")
print("File path:", output_path)
print("Shape:", saved_df.shape)
print("Columns:", list(saved_df.columns))


# Final verification

print("Final Promotion Master verification")
print("-----------------------------------")
print("Rows:", len(saved_df))
print(
    "Duplicate IDs:",
    saved_df["promotion_id"].duplicated().sum()
)
print(
    "Unique IDs:",
    saved_df["promotion_id"].nunique()
)
print(
    "Missing values:",
    saved_df.isnull().sum().sum()
)

print(
    "\nPromotion Master generation completed successfully."
)
```

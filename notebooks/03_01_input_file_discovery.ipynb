```python
from pathlib import Path
import pandas as pd
```

```python
current_path = Path.cwd()

project_root = current_path

while project_root != project_root.parent:
    if (project_root / "notebooks").exists() and (project_root / "data").exists():
        break
    project_root = project_root.parent

if not (
    (project_root / "notebooks").exists()
    and (project_root / "data").exists()
):
    raise FileNotFoundError("Could not locate the Retail Sales Analytics project root.")

PROJECT_ROOT = project_root
INCOMING_DIR = PROJECT_ROOT / "incoming"

INCOMING_DIR.mkdir(parents=True, exist_ok=True)

print("Project root:", PROJECT_ROOT)
print("Incoming folder:", INCOMING_DIR)
print("Incoming folder exists:", INCOMING_DIR.exists())
```

```python
SUPPORTED_EXTENSIONS = {".csv", ".xlsx", ".xls"}

input_files = [
    file
    for file in INCOMING_DIR.iterdir()
    if file.is_file() and file.suffix.lower() in SUPPORTED_EXTENSIONS
]

print("Files discovered:", len(input_files))

for file in input_files:
    print(file.name)
```

```python
def identify_dataset(file_name):
    name = file_name.lower()

    if "customer" in name:
        return "customers"

    if "product" in name:
        return "products"

    if "store" in name:
        return "stores"

    if "salesperson" in name:
        return "salespersons"

    if "promotion" in name:
        return "promotions"

    if "target" in name:
        return "targets"

    if "return" in name:
        return "returns"

    if "sales" in name:
        return "sales"

    return "unknown"
```

```python
test_file_names = [
    "customers_master.csv",
    "products_master.csv",
    "stores_master.csv",
    "salespersons_master.csv",
    "promotions_master.csv",
    "targets.csv",
    "sales_2024-01.csv",
    "returns_2024-01.csv",
    "unknown_file.csv"
]

for file_name in test_file_names:
    dataset = identify_dataset(file_name)
    print(f"{file_name} -> {dataset}")
```

```python
discovered_files = []

for file in input_files:
    dataset = identify_dataset(file.name)

    discovered_files.append({
        "file_name": file.name,
        "file_path": str(file),
        "extension": file.suffix.lower(),
        "dataset": dataset
    })

discovery_df = pd.DataFrame(discovered_files)

discovery_df
```

```python
import re

APPROVED_DATASETS = {
    "customers",
    "products",
    "stores",
    "salespersons",
    "promotions",
    "targets",
    "sales",
    "returns"
}

FILE_NAME_PATTERN = re.compile(
    r"^(customers|products|stores|salespersons|promotions|targets|sales|returns)_\d{4}_\d{2}\.csv$",
    re.IGNORECASE
)

print("Approved datasets:", sorted(APPROVED_DATASETS))
print("Filename pattern:", FILE_NAME_PATTERN.pattern)
```

```python
from src.config import RAW_DATA_PATH, PROJECT_ROOT

print("RAW_DATA_PATH:", RAW_DATA_PATH)
print("PROJECT_ROOT:", PROJECT_ROOT)
```

```python
from pathlib import Path

# Use the configured input folder
INPUT_FOLDER = Path(RAW_DATA_PATH)

# Find files currently present in the input folder
input_files = [
    file_path
    for file_path in INPUT_FOLDER.iterdir()
    if file_path.is_file()
]

valid_files = []
invalid_files = []

for file_path in input_files:
    file_name = file_path.name

    if FILE_NAME_PATTERN.fullmatch(file_name):
        valid_files.append(file_name)
    else:
        invalid_files.append(file_name)

print("Input folder:", INPUT_FOLDER)
print()

print("VALID FILES:")
if valid_files:
    for file_name in valid_files:
        print("  ACCEPTED:", file_name)
else:
    print("  None")

print()
print("INVALID FILES:")
if invalid_files:
    for file_name in invalid_files:
        print("  WOULD BE QUARANTINED:", file_name)
else:
    print("  None")

print()
print("Validation summary")
print("------------------")
print("Total files      :", len(input_files))
print("Valid files      :", len(valid_files))
print("Invalid files    :", len(invalid_files))
```

```python
from pathlib import Path
import shutil

QUARANTINE_FOLDER = PROJECT_ROOT / "data" / "quarantine"
QUARANTINE_FOLDER.mkdir(parents=True, exist_ok=True)

for file_name in invalid_files:
    source_path = INPUT_FOLDER / file_name
    destination_path = QUARANTINE_FOLDER / file_name

    if destination_path.exists():
        raise FileExistsError(
            f"Quarantine destination already exists: {destination_path}"
        )

    shutil.move(str(source_path), str(destination_path))

    print("QUARANTINED:", file_name)
    print("FROM       :", source_path)
    print("TO         :", destination_path)
```

```python
# Verify quarantine result

source_path = INPUT_FOLDER / "retail_sales.csv"
quarantine_path = QUARANTINE_FOLDER / "retail_sales.csv"

print("Quarantine verification")
print("-----------------------")
print("Still in input folder :", source_path.exists())
print("In quarantine folder  :", quarantine_path.exists())

if not source_path.exists() and quarantine_path.exists():
    print()
    print("PASS: Invalid file was successfully quarantined.")
else:
    print()
    print("FAIL: Quarantine verification requires review.")
```

```python
# Tracker 02.02 — Products Master Generator

import pandas as pd
import numpy as np
from faker import Faker
import random
from datetime import datetime, timedelta
```

```python
# Initialize Faker and random seed

fake = Faker("en_IN")

Faker.seed(42)
random.seed(42)
np.random.seed(42)

NUM_PRODUCTS = 2500

print("Faker initialized successfully.")
print("Products to generate:", NUM_PRODUCTS)
```

```python
# Generate Product IDs

product_ids = [
    f"PROD{str(i).zfill(6)}"
    for i in range(1, NUM_PRODUCTS + 1)
]

print("Total Product IDs:", len(product_ids))
print("First 5 IDs:", product_ids[:5])
print("Last 5 IDs:", product_ids[-5:])
```

```python
# Product categories, subcategories, brands and GST mapping

product_catalog = {
    "Electronics": {
        "subcategories": ["Mobile", "Laptop", "Headphones", "Smart TV", "Smart Watch"],
        "brands": ["Samsung", "Apple", "Sony", "LG", "OnePlus"],
        "gst": 18
    },
    "Grocery": {
        "subcategories": ["Rice", "Wheat Flour", "Cooking Oil", "Sugar", "Tea"],
        "brands": ["Aashirvaad", "Fortune", "Tata", "India Gate", "Organic India"],
        "gst": 5
    },
    "Fashion": {
        "subcategories": ["Shirt", "Jeans", "Shoes", "Kurti", "Jacket"],
        "brands": ["Nike", "Adidas", "Levis", "Puma", "Allen Solly"],
        "gst": 12
    },
    "Home & Kitchen": {
        "subcategories": ["Mixer", "Cookware", "Sofa", "Dining Table", "Curtains"],
        "brands": ["Prestige", "Philips", "IKEA", "Milton", "Cello"],
        "gst": 18
    },
    "Beauty & Personal Care": {
        "subcategories": ["Shampoo", "Soap", "Face Wash", "Perfume", "Moisturizer"],
        "brands": ["Dove", "Nivea", "Lakme", "L'Oreal", "Himalaya"],
        "gst": 18
    }
}

print("Categories:", list(product_catalog.keys()))
print("Total Categories:", len(product_catalog))
```

```python
# Generate product records

products = []

for product_id in product_ids:

    # Select category
    category = random.choice(list(product_catalog.keys()))

    # Select details based on category
    subcategory = random.choice(product_catalog[category]["subcategories"])
    brand = random.choice(product_catalog[category]["brands"])
    gst_percent = product_catalog[category]["gst"]

    # Generate pricing
    cost_price = round(np.random.uniform(100, 50000), 2)
    profit_margin_pct = round(np.random.uniform(10, 40), 2)

    unit_price = round(
        cost_price * (1 + profit_margin_pct / 100),
        2
    )

    # Generate remaining product details
    supplier_name = fake.company()
    stock_quantity = random.randint(10, 1000)

    launch_date = fake.date_between(
        start_date="-5y",
        end_date="today"
    )

    status = random.choices(
        ["Active", "Discontinued"],
        weights=[90, 10],
        k=1
    )[0]

    weight_kg = round(np.random.uniform(0.1, 25), 2)

    rating = round(
        np.random.uniform(3.0, 5.0),
        1
    )

    warranty_months = random.choice(
        [0, 6, 12, 24, 36]
    )

    product_name = f"{brand} {subcategory}"

    products.append({
        "product_id": product_id,
        "product_name": product_name,
        "category": category,
        "subcategory": subcategory,
        "brand": brand,
        "supplier_name": supplier_name,
        "unit_price": unit_price,
        "cost_price": cost_price,
        "profit_margin_pct": profit_margin_pct,
        "gst_percent": gst_percent,
        "stock_quantity": stock_quantity,
        "launch_date": launch_date,
        "status": status,
        "weight_kg": weight_kg,
        "rating": rating,
        "warranty_months": warranty_months
    })

print("Total product records generated:", len(products))
```

```python
# Convert product records into a Pandas DataFrame

products_df = pd.DataFrame(products)

print("DataFrame Shape:", products_df.shape)
print("Total Products:", len(products_df))
print("Total Columns:", len(products_df.columns))

products_df.head()
```

```python
# Validate product data

print("Missing values:")
print(products_df.isnull().sum())

print("\nDuplicate Product IDs:")
print(products_df["product_id"].duplicated().sum())

print("\nProduct ID uniqueness:")
print(products_df["product_id"].nunique())

print("\nPrice validation:")
print("Minimum Cost Price:", products_df["cost_price"].min())
print("Minimum Unit Price:", products_df["unit_price"].min())

print("\nProduct Status:")
print(products_df["status"].value_counts())
```

```python
# Save Products Master dataset

output_path = "../data/synthetic/products_master.csv"

products_df.to_csv(output_path, index=False)

print("Products Master saved successfully.")
print("File path:", output_path)
print("Rows saved:", len(products_df))
```

```python
# Confirm saved Product Master file

saved_df = pd.read_csv(output_path)

print("Saved file loaded successfully.")
print("Shape:", saved_df.shape)
print("Columns:", list(saved_df.columns))
```

```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path("../data/synthetic")

master_files = {
    "customers_master": ("customers_master.csv", 10000),
    "products_master": ("products_master.csv", 2500),
    "stores_master": ("stores_master.csv", 60),
    "salespersons_master": ("salespersons_master.csv", 500),
    "promotions_master": ("promotions_master.csv", 72),
}

print("=" * 70)
print("02.31 - STEP 2 - MASTER DATA VOLUME VALIDATION")
print("=" * 70)

results = []

for dataset, (filename, expected_rows) in master_files.items():

    file_path = BASE_DIR / filename
    df = pd.read_csv(file_path)

    actual_rows = len(df)
    status = "VALID" if actual_rows == expected_rows else "NEEDS REVIEW"

    results.append({
        "dataset": dataset,
        "expected_rows": expected_rows,
        "actual_rows": actual_rows,
        "status": status
    })

    print(f"\n{dataset}")
    print(f"Expected rows : {expected_rows:,}")
    print(f"Actual rows   : {actual_rows:,}")
    print(f"Status        : {status}")

report_df = pd.DataFrame(results)

if (report_df["status"] == "VALID").all():
    print("\n02.31 Step 2 validation: PASSED")
else:
    print("\n02.31 Step 2 validation: NEEDS REVIEW")
```

```python
from pathlib import Path
import pandas as pd

print("=" * 70)
print("02.31 - STEP 3 - TRANSACTION RECORD VOLUME VALIDATION")
print("=" * 70)

DATA_DIR = Path("../data/synthetic")

EXPECTED_SALES = 2_118_200
EXPECTED_RETURNS = 118_400
EXPECTED_TARGETS = 8_980


def count_csv_rows(file_path):
    total = 0

    for chunk in pd.read_csv(file_path, chunksize=100_000):
        total += len(chunk)

    return total


# --------------------------------------------------
# SALES
# --------------------------------------------------

sales_files = sorted(DATA_DIR.glob("sales_*.csv"))

actual_sales = 0

for file in sales_files:
    actual_sales += count_csv_rows(file)

print("\nSALES")
print("-" * 50)
print(f"Files checked   : {len(sales_files)}")
print(f"Expected rows   : {EXPECTED_SALES:,}")
print(f"Actual rows     : {actual_sales:,}")
print(f"Status          : {'VALID' if actual_sales == EXPECTED_SALES else 'NEEDS REVIEW'}")


# --------------------------------------------------
# RETURNS
# --------------------------------------------------

return_files = sorted(DATA_DIR.glob("returns_*.csv"))

actual_returns = 0

for file in return_files:
    actual_returns += count_csv_rows(file)

print("\nRETURNS")
print("-" * 50)
print(f"Files checked   : {len(return_files)}")
print(f"Expected rows   : {EXPECTED_RETURNS:,}")
print(f"Actual rows     : {actual_returns:,}")
print(f"Status          : {'VALID' if actual_returns == EXPECTED_RETURNS else 'NEEDS REVIEW'}")


# --------------------------------------------------
# TARGETS
# --------------------------------------------------

targets_file = DATA_DIR / "targets.csv"

actual_targets = len(pd.read_csv(targets_file))

print("\nTARGETS")
print("-" * 50)
print(f"Expected rows   : {EXPECTED_TARGETS:,}")
print(f"Actual rows     : {actual_targets:,}")
print(f"Status          : {'VALID' if actual_targets == EXPECTED_TARGETS else 'NEEDS REVIEW'}")


# --------------------------------------------------
# FINAL VALIDATION
# --------------------------------------------------

all_valid = (
    actual_sales == EXPECTED_SALES
    and actual_returns == EXPECTED_RETURNS
    and actual_targets == EXPECTED_TARGETS
)

print("\n" + "=" * 70)

if all_valid:
    print("02.31 Step 3 validation: PASSED")
else:
    print("02.31 Step 3 validation: NEEDS REVIEW")
```

```python
from pathlib import Path

print("02.26 STEP 2 - PREPARE CORRECTION FILE OUTPUT")
print("=" * 65)

BASE_DIR = Path.cwd().parent
SALES_DIR = BASE_DIR / "data" / "synthetic"

SOURCE_FILE = SALES_DIR / "sales_2026-06.csv"

OUTPUT_DIR = SALES_DIR / "correction_file_adjusted"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

CORRECTION_FILE = OUTPUT_DIR / "sales_2026-06_CORRECTION.csv"

print("\nSOURCE FILE")
print("-" * 65)
print(SOURCE_FILE)

print("\nCORRECTION OUTPUT FOLDER")
print("-" * 65)
print(OUTPUT_DIR)

print("\nCORRECTION FILE")
print("-" * 65)
print(CORRECTION_FILE)

print("\nOriginal file exists:", SOURCE_FILE.exists())
print("Output folder exists:", OUTPUT_DIR.exists())

if SOURCE_FILE.exists() and OUTPUT_DIR.exists():
    print("\n02.26 Step 2 preparation: PASSED")
else:
    print("\n02.26 Step 2 preparation: NEEDS REVIEW")
```

```python
import pandas as pd
import numpy as np

print("02.26 STEP 3 - CREATE CORRECTION FILE")
print("=" * 65)

# Load the original June 2026 file
sales = pd.read_csv(SOURCE_FILE)

# Keep the original values for comparison
original_sales = sales.copy()

# Number of records to correct
CORRECTION_ROWS = 300

# Reproducible selection
rng = np.random.default_rng(2026)

correction_indices = rng.choice(
    sales.index,
    size=CORRECTION_ROWS,
    replace=False
)

# Apply controlled corrections
sales.loc[correction_indices, "quantity"] = (
    sales.loc[correction_indices, "quantity"] + 1
)

sales.loc[correction_indices, "discount_pct"] = (
    sales.loc[correction_indices, "discount_pct"] + 1
).clip(upper=50)

# Save corrected file
sales.to_csv(
    CORRECTION_FILE,
    index=False
)

print("\nCORRECTION SUMMARY")
print("-" * 65)

print("Original rows:", len(original_sales))
print("Corrected rows:", len(sales))
print("Rows selected for correction:", len(correction_indices))

print("\nCORRECTED COLUMNS")
print("-" * 65)
print("quantity")
print("discount_pct")

print("\nIDENTIFIER PRESERVATION")
print("-" * 65)

sales_line_id_preserved = (
    original_sales["sales_line_id"].equals(
        sales["sales_line_id"]
    )
)

order_id_preserved = (
    original_sales["order_id"].equals(
        sales["order_id"]
    )
)

print("sales_line_id preserved:", sales_line_id_preserved)
print("order_id preserved:", order_id_preserved)

print("\nOUTPUT")
print("-" * 65)
print("Correction file:", CORRECTION_FILE)
print("File exists:", CORRECTION_FILE.exists())

if (
    len(sales) == len(original_sales)
    and sales_line_id_preserved
    and order_id_preserved
    and CORRECTION_FILE.exists()
):
    print("\n02.26 Step 3 correction creation: PASSED")
else:
    print("\n02.26 Step 3 correction creation: NEEDS REVIEW")
```

```python
import pandas as pd

print("02.26 STEP 4 - FINAL CORRECTION FILE VALIDATION")
print("=" * 65)

# Load files
original = pd.read_csv(SOURCE_FILE)
corrected = pd.read_csv(CORRECTION_FILE)

print("\nFILE VALIDATION")
print("-" * 65)

print("Original rows:", len(original))
print("Corrected rows:", len(corrected))
print("Original columns:", len(original.columns))
print("Corrected columns:", len(corrected.columns))

# --------------------------------------------------
# 1. Identifier validation
# --------------------------------------------------

sales_line_id_preserved = (
    original["sales_line_id"].tolist()
    == corrected["sales_line_id"].tolist()
)

order_id_preserved = (
    original["order_id"].tolist()
    == corrected["order_id"].tolist()
)

print("\nIDENTIFIER VALIDATION")
print("-" * 65)

print("sales_line_id preserved:", sales_line_id_preserved)
print("order_id preserved:", order_id_preserved)

# --------------------------------------------------
# 2. Correction validation
# --------------------------------------------------

quantity_changes = (
    original["quantity"] != corrected["quantity"]
).sum()

discount_changes = (
    original["discount_pct"] != corrected["discount_pct"]
).sum()

print("\nCORRECTION VALIDATION")
print("-" * 65)

print("Quantity changes:", quantity_changes)
print("Discount changes:", discount_changes)

# --------------------------------------------------
# 3. Check ONLY non-correction columns
# --------------------------------------------------

correction_columns = ["quantity", "discount_pct"]

other_columns = [
    col for col in original.columns
    if col not in correction_columns
]

unexpected_changes = 0

for col in other_columns:

    original_values = original[col].fillna("__MISSING__").astype(str)
    corrected_values = corrected[col].fillna("__MISSING__").astype(str)

    unexpected_changes += (
        original_values != corrected_values
    ).sum()

print("\nNON-CORRECTION COLUMN VALIDATION")
print("-" * 65)

print("Unexpected changes outside correction columns:",
      unexpected_changes)

# --------------------------------------------------
# 4. Final validation
# --------------------------------------------------

final_validation = (
    len(original) == len(corrected)
    and len(original.columns) == len(corrected.columns)
    and sales_line_id_preserved
    and order_id_preserved
    and quantity_changes == 300
    and discount_changes == 300
    and unexpected_changes == 0
)

print("\nFINAL VALIDATION")
print("-" * 65)

print("Row count preserved:", len(original) == len(corrected))
print("Column count preserved:", len(original.columns) == len(corrected.columns))
print("Identifiers preserved:",
      sales_line_id_preserved and order_id_preserved)
print("Exactly 300 quantity corrections:",
      quantity_changes == 300)
print("Exactly 300 discount corrections:",
      discount_changes == 300)
print("No unexpected column changes:",
      unexpected_changes == 0)

if final_validation:
    print("\n02.26 Step 4 validation: PASSED")
else:
    print("\n02.26 Step 4 validation: NEEDS REVIEW")
```

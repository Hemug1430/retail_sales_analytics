```python
# Tracker 02.03 — Store Master Generator

import pandas as pd
import numpy as np
from faker import Faker
import random
from datetime import datetime, timedelta
```

```python
# Initialize Faker and random seed

fake = Faker("en_IN")

Faker.seed(42)
random.seed(42)
np.random.seed(42)

NUM_STORES = 60

print("Faker initialized successfully.")
print("Stores to generate:", NUM_STORES)
```

```python
# Generate Store IDs

store_ids = [
    f"STORE{str(i).zfill(5)}"
    for i in range(1, NUM_STORES + 1)
]

print("Total Store IDs:", len(store_ids))
print("First 5 IDs:", store_ids[:5])
print("Last 5 IDs:", store_ids[-5:])
```

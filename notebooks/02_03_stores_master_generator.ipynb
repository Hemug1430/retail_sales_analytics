```python
# Tracker 02.03 — Store Master Generator

import pandas as pd
import numpy as np
from faker import Faker
import random
from datetime import datetime, timedelta
```

```python
# Initialize Faker and random seed

fake = Faker("en_IN")

Faker.seed(42)
random.seed(42)
np.random.seed(42)

NUM_STORES = 60

print("Faker initialized successfully.")
print("Stores to generate:", NUM_STORES)
```

```python
# Generate Store IDs

store_ids = [
    f"STORE{str(i).zfill(3)}"
    for i in range(1, NUM_STORES + 1)
]

print("Total Store IDs:", len(store_ids))
print("First 5 IDs:", store_ids[:5])
print("Last 5 IDs:", store_ids[-5:])
```

```python
# Define approved store regions, cities and store type distribution

region_store_counts = {
    "South": 18,
    "West": 14,
    "North": 13,
    "East": 9,
    "Central": 6
}

region_city_pool = {
    "South": [
        "Hyderabad",
        "Bengaluru",
        "Chennai",
        "Kochi",
        "Vijayawada",
        "Visakhapatnam",
        "Coimbatore"
    ],
    "West": [
        "Mumbai",
        "Pune",
        "Ahmedabad",
        "Surat",
        "Nagpur",
        "Nashik",
        "Goa"
    ],
    "North": [
        "Delhi",
        "Gurugram",
        "Noida",
        "Jaipur",
        "Chandigarh",
        "Lucknow",
        "Ludhiana"
    ],
    "East": [
        "Kolkata",
        "Bhubaneswar",
        "Patna",
        "Ranchi",
        "Guwahati",
        "Siliguri"
    ],
    "Central": [
        "Indore",
        "Bhopal",
        "Raipur",
        "Jabalpur",
        "Gwalior"
    ]
}

store_types = {
    "Mall": 0.30,
    "High Street": 0.27,
    "Neighborhood": 0.30,
    "Outlet": 0.13
}

print("Approved region distribution:", region_store_counts)
print("Total stores:", sum(region_store_counts.values()))
print("Approved store types:", store_types)
```

```python
city_to_state = {
    "Hyderabad": "Telangana",
    "Bengaluru": "Karnataka",
    "Chennai": "Tamil Nadu",
    "Kochi": "Kerala",
    "Vijayawada": "Andhra Pradesh",
    "Visakhapatnam": "Andhra Pradesh",
    "Coimbatore": "Tamil Nadu",
    "Mumbai": "Maharashtra",
    "Pune": "Maharashtra",
    "Ahmedabad": "Gujarat",
    "Surat": "Gujarat",
    "Nagpur": "Maharashtra",
    "Nashik": "Maharashtra",
    "Goa": "Goa",
    "Delhi": "Delhi",
    "Gurugram": "Haryana",
    "Noida": "Uttar Pradesh",
    "Jaipur": "Rajasthan",
    "Chandigarh": "Chandigarh",
    "Lucknow": "Uttar Pradesh",
    "Ludhiana": "Punjab",
    "Kolkata": "West Bengal",
    "Bhubaneswar": "Odisha",
    "Patna": "Bihar",
    "Ranchi": "Jharkhand",
    "Guwahati": "Assam",
    "Siliguri": "West Bengal",
    "Indore": "Madhya Pradesh",
    "Bhopal": "Madhya Pradesh",
    "Raipur": "Chhattisgarh",
    "Jabalpur": "Madhya Pradesh",
    "Gwalior": "Madhya Pradesh"
}
```

```python
# Generate store records
regions = [r for r, cnt in region_store_counts.items() for _ in range(cnt)]
random.shuffle(regions)

stores = []
for idx, store_id in enumerate(store_ids):
    region = regions[idx % len(regions)]
    city = random.choice(region_city_pool[region])
    state = city_to_state[city]
    store_type = random.choices(list(store_types.keys()), weights=list(store_types.values()), k=1).pop()
    opening_date = fake.date_between(start_date='-8y', end_date='-6m')
    status = random.choices(['Active', 'Inactive'], weights=[0.92, 0.08], k=1).pop()
    stores.append({
        'store_id': store_id,
        'store_name': f'{city} {store_type} Store {idx+1:02d}',
        'store_type': store_type,
        'city': city,
        'state': state,
        'region': region,
        'opening_date': str(opening_date),
        'floor_area_sqft': random.randint(1500, 12000), 'status': status
    })

print('Generated stores:', len(stores))
```

```python
## Convert store records into a Pandas DataFrame

stores_df = pd.DataFrame(stores)

print("DataFrame Shape:", stores_df.shape)
print("Total Stores:", len(stores_df))
print("Total Columns:", len(stores_df.columns))

stores_df.head()
```

```python
# Validate store data

print("Missing values:")
print(stores_df.isnull().sum())

print("\nDuplicate Store IDs:")
print(stores_df["store_id"].duplicated().sum())

print("\nStore ID uniqueness:")
print(stores_df["store_id"].nunique())

print("\nStore Status:")
print(stores_df["status"].value_counts())

print("\nStore Type:")
print(stores_df["store_type"].value_counts())
```

```python
# Save Stores Master dataset

from pathlib import Path

output_path = Path("../data/synthetic/stores_master.csv")

output_path.parent.mkdir(parents=True, exist_ok=True)

stores_df.to_csv(output_path, index=False)

print("Stores Master saved successfully.")
print("File path:", output_path)
print("Rows saved:", len(stores_df))
```

```python
# Confirm saved Stores Master file

saved_df = pd.read_csv(output_path)

print("Saved file loaded successfully.")
print("Shape:", saved_df.shape)
print("Columns:", list(saved_df.columns))
```
